#🔧 Transformação

```python
# Achatar a estrutura JSON

df = pd.json_normalize(df.to_dict(orient = 'records'))
df.head()
```

```python
df.info()
```

```python
# Renomeando as colunas

df.columns = [col.split('.')[-1] for col in df.columns]
df.info()
```

```python
# Trocando a vírgula para poder transformar em float
df['Total'] = df['Total'].astype(str).str.replace(',', '.').str.strip()

# Tirando os espacos
df['Total'] = df['Total'].str.replace(' ', '')
df['Total']

# Converter os vazios para NaN
df['Total'] = pd.to_numeric(df['Total'], errors='coerce')

# Trocar os NaN por 0
df['Total'] = df['Total'].fillna(0)
```

```python
# Transformando em float

df['Total'] = df['Total'].astype(float)
df['Total']
```

```python
# Deixando as linhas em minúsculo

df = df.applymap(lambda s: s.lower() if type(s) == str else s)
df.head()
```

```python
# Retirando caracteres indesejados

df = df.replace(r'[^a-zA-Z0-9 ]', '', regex = True)
df.head()
```

```python
# Ajustando coluna

df['Contract'] = df['Contract'].replace('monthtomonth', 'month')
df['Contract']
```

```python
# Adicionando a coluna de contas diarias

df['Contas_Diarias'] = df['Monthly'] / 30
df['Contas_Diarias']
```

```python
# Deixando as colunas de dinheiro com cifrao e somente com 2 casas decimais

colunas_dinheiro = ['Monthly', 'Total', 'Contas_Diarias']

for col in colunas_dinheiro:
    # Limpar os dados
    df[col] = df[col].astype(str).str.replace(' ', '').str.replace(',', '.')

    # Tranformar em dados númericos
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

    # Formatar
    df[col] = df[col].apply(lambda x: f'R${x:.2f}')

    df[col] = df[col].str.replace('R$', '').str.replace(',', '.').astype(float)


```

```python
# Renomeando e padronizando as colunas

# Dicionário de tradução e padronização
colunas_novas = {
    'customerID': 'id_cliente',
    'Churn': 'evasao',
    'gender': 'genero',
    'SeniorCitizen': 'idoso',
    'Partner': 'tem_parceiro',
    'Dependents': 'tem_dependentes',
    'tenure': 'meses_contrato',
    'PhoneService': 'servico_telefone',
    'MultipleLines': 'multiplas_linhas',
    'InternetService': 'servico_internet',
    'OnlineSecurity': 'seguranca_online',
    'OnlineBackup': 'backup_online',
    'DeviceProtection': 'protecao_dispositivo',
    'TechSupport': 'suporte_tecnico',
    'StreamingTV': 'streaming_tv',
    'StreamingMovies': 'streaming_filmes',
    'Contract': 'tipo_contrato',
    'PaperlessBilling': 'fatura_digital',
    'PaymentMethod': 'metodo_pagamento',
    'Monthly': 'valor_mensal',
    'Total': 'valor_total',
    'Contas_Diarias': 'valor_diario'
}

df = df.rename(columns=colunas_novas)
df.head()
```

```python
# Selecionando colunas que serão booleanas

colunas_sim_nao = [
    'evasao', 'tem_parceiro', 'tem_dependentes', 'servico_telefone', 'multiplas_linhas',
    'seguranca_online', 'backup_online', 'protecao_dispositivo',
    'suporte_tecnico', 'streaming_tv', 'streaming_filmes', 'fatura_digital'
]

for col in colunas_sim_nao:
    # Isso transforma 'yes' em True e QUALQUER outra coisa em False
    df[col] = df[col].astype(str).str.lower().str.strip() == 'yes'

# Ajustando o Idoso para Booleano
df['idoso'] = df['idoso'] == 1
```

```python
total_evasao = df['evasao'].sum()
total_sem_evasao = df['evasao'].count() - total_evasao
```

```python
# Análise inicial de evasão

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

dados_evasao = pd.DataFrame({
    'Categoria': ['Ativos', 'Evasão'],
    'Quantidade': [total_sem_evasao, total_evasao]
})

def func_personalizada(pct, allvals):
    absolute = int(round(pct / 100. * sum(allvals)))
    return f'{absolute}\n({pct:.1f}%)'

cores = sns.color_palette("husl", 2)

plt.figure(figsize=(8, 8))
plt.pie(
    dados_evasao['Quantidade'],
    labels=dados_evasao['Categoria'],
    colors=cores,
    startangle=90,
    autopct=lambda pct: func_personalizada(pct, dados_evasao['Quantidade']),
    explode=(0, 0.08),
    shadow=True,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5}
)

plt.title(
    'Distribuição de Clientes Ativos e em Evasão',
    fontsize=18,
    loc='left',
    pad=20,
    fontweight='bold'
)

plt.axis('equal')
plt.show()

plt.savefig('evasao.png')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

tradutor_evasao = {True: 'Sim', False: 'Não'}
tradutor_genero = {'male': 'Masculino', 'female': 'Feminino'}

df_plot = df.copy()
df_plot['genero_pt'] = df_plot['genero'].map(tradutor_genero)
df_plot['evasao_pt'] = df_plot['evasao'].map(tradutor_evasao)


plt.figure(figsize=(10, 8))

ax = sns.histplot(
    data=df_plot,
    x='genero_pt',
    hue='evasao_pt',
    multiple='stack',
    palette={'Sim': '#E74C3C', 'Não': '#2ECC71'},
    shrink=0.7
)

for container in ax.containers:
    ax.bar_label(container, label_type='center', color='white', fontweight='bold', fontsize=12)

plt.title('Total de Clientes por Gênero vs. Cancelamentos', fontsize=16, fontweight='bold', loc='left', pad=20)
plt.xlabel('Gênero', fontsize=12)
plt.ylabel('Quantidade Total de Clientes', fontsize=12)
plt.legend(title='Evadiu?', labels=['Não', 'Sim'])

sns.despine()
plt.show()

plt.savefig('evasao_genero.png')
```

```python
# Análise inicial evasão por categorias

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

tradutor_bool = {True: 'Sim', False: 'Não'}
colunas_bool = ['tem_parceiro', 'tem_dependentes', 'servico_telefone',
                'seguranca_online', 'suporte_tecnico', 'multiplas_linhas',
                'streaming_tv', 'streaming_filmes', 'fatura_digital']

fig, axs = plt.subplots(3, 3, figsize=(22, 18))
fig.subplots_adjust(hspace=0.4, wspace=0.3)
fig.suptitle('Análise de Evasão por Categoria (Sim/Não)', fontsize=25, fontweight='bold', y=0.95)

for i, col in enumerate(colunas_bool):
  ax = axs.flat[i]

  df_temp = df.copy()
  df_temp[col] = df_temp[col].map(tradutor_bool)
  df_temp['evasao_pt'] = df_temp['evasao'].map(tradutor_bool)

  sns.histplot(
      data = df_temp,
      x = col,
      hue = 'evasao_pt',
      hue_order = ['Não', 'Sim'],
      multiple = 'stack',
      ax= ax,
      palette = {'Sim': '#C0392B', 'Não': '#27AE60'},
      shrink = 0.8
  )

  for container in ax.containers:
    ax.bar_label(container, padding = 3, fontweight = 'bold')

  titulo_limpo = col.replace('_', ' ').capitalize()
  ax.set_title(f'Evasão por {titulo_limpo}', fontsize=15, fontweight='bold', loc='left')
  ax.set_ylabel('Quantidade de Clientes')
  ax.set_xlabel(f'{titulo_limpo}?')
  ax.legend(title='Evadiu?', labels = ['Não', 'Sim'])
  ax.tick_params(axis='both', labelsize=11)
  ax.grid(axis='y', linestyle='--', alpha=0.3)

  sns.despine(ax=ax)

handles, labels = ax.get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    title='Evadiu?',
    loc='lower center',
    ncol=2,
    fontsize=12,
    title_fontsize=13
)

plt.show()

plt.savefig('analise_categoria.png')

```

```python
# Análise proporção de evasão

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

categoria = ['tem_parceiro', 'tem_dependentes', 'servico_telefone',
                'seguranca_online', 'suporte_tecnico', 'multiplas_linhas',
                'streaming_tv', 'streaming_filmes', 'fatura_digital']

quantidade = []

for col in categoria:
  qtd = df[
      (df['evasao'] == True) & (df[col] == True)
  ].shape[0]
  quantidade.append(qtd)

df_evasao = pd.DataFrame({
    'Categoria': categoria,
    'Quantidade': quantidade
})

df_evasao = df_evasao.sort_values(by='Quantidade', ascending=False)

cores = sns.color_palette("husl", len(df_evasao))
explode = [0.1 if i == 0 else 0 for i in range(len(df_evasao))]
total = df_evasao['Quantidade'].sum()

labels_legiveis = {
    'tem_parceiro': 'Possui parceiro',
    'tem_dependentes': 'Possui dependentes',
    'servico_telefone': 'Serviço de telefone',
    'seguranca_online': 'Segurança online',
    'suporte_tecnico': 'Suporte técnico',
    'multiplas_linhas': 'Múltiplas linhas',
    'streaming_tv': 'Streaming de TV',
    'streaming_filmes': 'Streaming de filmes',
    'fatura_digital': 'Fatura digital'
}

df_evasao['Categoria_label'] = df_evasao['Categoria'].map(labels_legiveis)

def autopct_format(pct):
    valor = int(round(pct * total / 100))
    return f'{valor}\n({pct:.1f}%)'

plt.figure(figsize=(9, 9))
plt.pie(
    df_evasao['Quantidade'],
    labels=df_evasao['Categoria_label'],
    colors=cores,
    startangle=90,
    autopct=autopct_format,
    explode=explode,
    shadow=True,
    wedgeprops={'edgecolor':'white', 'linewidth': 1}
)

plt.title(
    'Distribuição dos clientes evadidos com a categoria de serviço',
    fontsize=18,
    loc='left',
    pad=20,
    fontweight='bold'
)

plt.show()

plt.savefig('evasao_por_categoria.png')


```

```python
# Análise de evasão para categorias numéricas

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

colunas_valor = ['valor_mensal', 'valor_total', 'valor_diario']

for col in colunas_valor:
  df[f'faixa_{col}'] = pd.qcut(df[col], q = 5)

resultados = {}

for col in colunas_valor:
  taxa = (
      df.groupby(f'faixa_{col}')['evasao']
      .mean()
      .reset_index()
  )
  resultados[col] = taxa

fig = plt.figure(figsize=(14, 10))
plt.title('Evasão por Faixa de Valor', fontsize=18, fontweight='bold', loc='left', pad=20)
gs = fig.add_gridspec(2, 2)

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, :])

# Valor mensal
sns.barplot(
    data=resultados['valor_mensal'],
    x='faixa_valor_mensal',
    y='evasao',
    ax=ax1,
    color = '#030D4F'
)
ax1.set_title('Evasão por Valor Mensal', fontsize = 18)
ax1.set_xlabel('Faixa de valor')
ax1.set_ylabel('Taxa de evasão')
ax1.tick_params(axis='x', rotation=45)
ax1.grid(axis='y', linestyle='--', alpha=0.3)

# Valor total
sns.barplot(
    data=resultados['valor_total'],
    x='faixa_valor_total',
    y='evasao',
    ax=ax2,
    color = '#FB0C06'
)
ax2.set_title('Evasão por Valor Total', fontsize = 18)
ax2.set_xlabel('Faixa de valor')
ax2.set_ylabel('')
ax2.tick_params(axis = 'x', rotation = 45)
ax2.grid(axis='y', linestyle='--', alpha=0.3)

# Valor diário
sns.barplot(
    data=resultados['valor_diario'],
    x='faixa_valor_diario',
    y='evasao',
    ax=ax3,
    color = '#FFC52C'
)
ax3.set_title('Evasão por Valor Diário', fontsize = 18)
ax3.set_xlabel('Faixa de valor')
ax3.set_ylabel('Taxa de evasão')
ax3.tick_params(axis='x', rotation=45)
ax3.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()

plt.show()

plt.savefig('evasao_por_valor.png')



```

#📄Relatorio Final

A evasão de clientes é um dos principais desafios da TelecomX, porque impacta diretamente o andamento da empresa e relação com o cliente. Nesse relatório, foram realizadas análises relacionadas a essa evasão a fim de identificar padrões e mitigar o problema

De primeiro modo, os dados da empresa foram tratados e limpados, de forma que foi feita a normalização do arquivo JSON a fim de melhorar a visualização da base de dados. Após isso, foi feita uma mudança de tipo de algumas colunas do dataframe, como o surgimento de colunas booleanas e float, além da padronização dos nomes das colunas e o tratamentos de colunas do tipo string

Após o tratamento dessas colunas, foram feitas análises visuais para identificar padrões em relação a evasão de clientes. Nessa parte do projeto, foram feitos gráficos que visavam relacionar a evasão com outros fatores, como o gênero, idade, valor que o cliente paga e entre outros. Cada gráfico foi feito a fim de deixar a análise mais fácil de ser visualizada

Dessas análises concluímos que:

---




**Análise de Evasão por Categoria (Sim/Não)**

*   A empresa possui um bom número de clientes ativos (aproximadamente 75%)

*   A evasão por gênero não possui grandes disparidades

*   Os clientes que tem parceiros evadem mais

*   Proporcionalmente, os clientes não possuem dependentes cancelam mais o serviço



---


**Distribuição dos clientes evadidos com a categoria de serviço**

*   Grande parte da evasão (mais de 50%) são de pessoas que tem Serviço de telefone, Fatura digital e Múltiplas linhas




---

**Evasão por Valor Total**

*   40% dos clientes que pagam o valor total até 264.60 evadem

*   35% dos clientes que pagam o valor mensal de 79.05 até 94.30 evadem

*   35% dos clientes que pagam o valor diário de 2.63 até 3.14 evadem mais



---

**Distribuição da Quantidade de Serviços por Evasão**
*   Nessa análise foi possível identificar que clientes que concentram mais serviços tendem a cancelar menos o serviço

---


Diante dos padrões identificados, recomenda-se que a TelecomX desenvolva estratégias de retenção direcionadas aos perfis de clientes que apresentam maior taxa de evasão, especialmente aqueles associados a faixas mais elevadas de valor relativo. A empresa pode investir em benefícios adicionais, atendimento diferenciado ou comunicação mais clara sobre os serviços oferecidos, a fim de aumentar a percepção de valor por parte desses clientes. Ademais, a revisão dos pacotes que agregam múltiplos serviços pode contribuir para a redução da evasão, seja por meio da simplificação das ofertas ou do reforço na divulgação de seus benefícios. Por fim, sugere-se a realização de análises mais aprofundadas, como abordagens multivariadas ou modelos preditivos, para compreender de forma mais abrangente os fatores que influenciam a evasão
